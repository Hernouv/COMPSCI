```python
#w11a4
def areaAnagrams(path1: str, path2: str) -> bool:
    with open(path1, 'r') as f1:
        s1 = f1.read()
    with open(path2, 'r') as f2:
        s2 = f2.read()

    def normalize(s):
        return ''.join(c.lower() for c in s if c.isalpha())

    s1_norm = normalize(s1)
    s2_norm = normalize(s2)
    
    return sorted(s1_norm) == sorted(s2_norm)
```

```python
#w11a8
def fibonacci_words(n: int, a: str = 'a', b: str = 'b'):
    if n < 0:
        return []
    
    result = []
    f0, f1 = a, b
    if n >= 0:
        result.append(f"f(0) = {f0}")
    if n >= 1:
        result.append(f"f(1) = {f1}")
    
    for i in range(2, n + 1):
        f_next = f1 + f0
        result.append(f"f({i}) = {f_next}")
        f0, f1 = f1, f_next
    
    with open('output.txt', 'w') as f:
        for line in result:
            f.write(line + '\n')

n = int(input())
fibonacci_words(n)
```

```python
#w11a10
def getMoney(path: str) -> list[int]:
    try:
        with open(path, 'r') as file:
            content = file.read().strip()
            if not content:
                return []
            return [int(x) for x in content.split()]
    except FileNotFoundError:
        return []
    except ValueError:
        return []
    except Exception as e:
        return []
    
def rob(houses: list[int]) -> int:
    if not houses:
        return 0
    if len(houses) == 1:
        return houses[0]

    dp = [0] * len(houses)
    dp[0] = houses[0]
    dp[1] = max(houses[0], houses[1])

    for i in range(2, len(houses)):
        dp[i] = max(houses[i] + dp[i-2], dp[i-1])

    return dp[-1]
```

```python
#w11a12
def maximumProduct(path: str) -> int:
    with open(path, 'r') as file:
        n_line = file.readline().strip()
        if not n_line:
            return 0
        n = int(n_line)
        
        numbers_line = file.readline().strip()
        if not numbers_line:
            return 0
        
        nums = list(map(int, numbers_line.split()))
        
        nums.sort()
        
        product1 = nums[-1] * nums[-2] * nums[-3]
        
        product2 = nums[0] * nums[1] * nums[-1]
            
        return max(product1, product2)
```

```python
#w11a13
def moveZeroes(fileName: str) -> list:
    with open(fileName, 'r') as file:
        n_line = file.readline().strip()
        if not n_line:
            return []
        n = int(n_line)
        
        numbers_line = file.readline().strip()
        if not numbers_line:
            return []
        
        nums = list(map(int, numbers_line.split()))
        
        if len(nums) > n:
            nums = nums[:n]
        
        non_zeros = [x for x in nums if x != 0]
        zeros = [0] * (len(nums) - len(non_zeros))
        result = non_zeros + zeros
        
        return result
    
```

```python
#w11a14
def averageTime(path: str) -> float:
    total_time = 0.0
    lap_count = 0
    
    with open(path, 'r') as file:
        lines = file.readlines()
        
        if len(lines) > 0:
            for line in lines[1:]:
                line = line.strip()
                if not line:
                    continue
                
                parts = line.split(',')
                if len(parts) >= 2:
                    try:
                        time_str = parts[1].strip()
                        time = float(time_str)
                        total_time += time
                        lap_count += 1
                    except ValueError:
                        continue
    
    if lap_count == 0:
        return 0.0
    
    average = total_time / lap_count
    return f"{average:.2f}"    
```

```python
#w11a15
def findMovies(path: str) -> list[str]:
    result = []
    
    with open(path, 'r') as file:
        lines = file.readlines()
        
        if len(lines) > 0:
            header = lines[0].strip().split(',')
            
            movie_idx = -1
            desc_idx = -1
            rating_idx = -1
            
            for i, col in enumerate(header):
                col = col.strip()
                if 'movie' in col.lower():
                    movie_idx = i
                elif 'description' in col.lower():
                    desc_idx = i
                elif 'rating' in col.lower():
                    rating_idx = i
            
            if movie_idx == -1 or desc_idx == -1 or rating_idx == -1:
                return ["All are boring!"]
            
            for line in lines[1:]:
                line = line.strip()
                if not line:
                    continue
                
                parts = line.split(',')
                if len(parts) >= 4:
                    try:
                        movie_name = parts[movie_idx].strip()
                        description = parts[desc_idx].strip().lower()
                        rating = float(parts[rating_idx].strip())
                        
                        if rating > 8.0 and 'boring' not in description:
                            result.append(movie_name)
                    except (ValueError, IndexError):
                        continue
    
    if not result:
        return ["All are boring!"]
    
    return result
    
```

```python
#w11a16
import math
def calculate_perimeter(path: str):
    try:
        with open(path, 'r', encoding='utf-8') as file:
            lines = file.readlines()
            
            i = 0
            while i < len(lines):
                line = lines[i].strip()
                
                if not line:
                    i += 1
                    continue
                
                if line.startswith("TYPE:"):
                    shape_type = line.split(":")[1].strip()
                    
                    if shape_type == "SQUARE":
                        i += 1
                        if i < len(lines):
                            side_line = lines[i].strip()
                            if side_line.startswith("SIDE:"):
                                side = float(side_line.split(":")[1].strip())
                                perimeter = 4 * side
                                print(f"{perimeter:.2f}")
                    
                    elif shape_type == "RECTANGLE":
                        i += 1
                        if i < len(lines):
                            dim_line = lines[i].strip()
                            if "WIDTH:" in dim_line and "HEIGHT:" in dim_line:
                                parts = dim_line.split()
                                width = 0
                                height = 0
                                
                                for part in parts:
                                    if part.startswith("WIDTH:"):
                                        width = float(part.split(":")[1])
                                    elif part.startswith("HEIGHT:"):
                                        height = float(part.split(":")[1])
                                
                                perimeter = 2 * (width + height)
                                print(f"{perimeter:.2f}")
                    
                    elif shape_type == "CIRCLE":
                        i += 1
                        if i < len(lines):
                            radius_line = lines[i].strip()
                            if radius_line.startswith("RADIUS:"):
                                radius = float(radius_line.split(":")[1].strip())
                                perimeter = 2 * math.pi * radius
                                print(f"{perimeter:.2f}")
                
                i += 1
                
    except FileNotFoundError:
        print("Không tìm thấy file!")
    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")
```

```python
#w11a17
def productExceptSelf(path: str) -> list[int]:
    try:
        with open(path, 'r', encoding='utf-8') as file:
            n_line = file.readline().strip()
            if not n_line:
                return []
            n = int(n_line)
            
            nums_line = file.readline().strip()
            if not nums_line:
                return []
            
            nums = list(map(int, nums_line.split()))
            if len(nums) != n:
                return []
            
            result = [1] * n
            
            left_product = 1
            for i in range(n):
                result[i] = left_product
                left_product *= nums[i]
            
            right_product = 1
            for i in range(n-1, -1, -1):
                result[i] *= right_product
                right_product *= nums[i]
            
            return result
            
    except FileNotFoundError:
        print(f"Không tìm thấy file: {path}")
        return []
    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")
        return []
```

```python
#w11a18
import math

def estimatedTime(path: str) -> float:
    try:
        src_x = src_y = des_x = des_y = velocity = 0
        
        with open(path, 'r') as f:
            for line in f:
                line = line.strip()
                if line.startswith("SRC"):
                    parts = line.replace(':', ' ').split()
                    src_x = float(parts[2])
                    src_y = float(parts[4])
                elif line.startswith("DES"):
                    parts = line.replace(':', ' ').split()
                    des_x = float(parts[2])
                    des_y = float(parts[4])
                elif line.startswith("VELOCITY"):
                    parts = line.split(':')
                    velocity = float(parts[1].split()[0])
        
        distance = math.hypot(des_x - src_x, des_y - src_y)
        return round(distance / velocity, 2)
        
    except:
        return 0.0
```

```python
#w11a20
def imageSmoother(path: str):
    try:
        with open(path, 'r', encoding='utf-8') as file:
            M = int(file.readline().strip())
            
            image = []
            for _ in range(M):
                row = list(map(int, file.readline().strip().split()))
                image.append(row)
            
            smoothed = [[0] * M for _ in range(M)]
            
            for i in range(M):
                for j in range(M):
                    total = 0
                    count = 0
                    
                    for di in [-1, 0, 1]:
                        for dj in [-1, 0, 1]:
                            ni, nj = i + di, j + dj
                            
                            if 0 <= ni < M and 0 <= nj < M:
                                total += image[ni][nj]
                                count += 1
                    
                    smoothed[i][j] = total // count
            
            for i in range(M):
                print(' '.join(map(str, smoothed[i])))
                
    except FileNotFoundError:
        print(f"Không tìm thấy file: {path}")
    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")
```
